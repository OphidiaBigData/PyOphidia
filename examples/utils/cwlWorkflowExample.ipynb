### Example of usage of PyOphidia for running an Ophidia CWL-based workflow

NOTE: to run this notebook PyOphidia should have been installed with *cwl* and *display* support. 

Connect to the Ophidia server side

```python
from PyOphidia import Cube, Client, Workflow, Experiment
cli = Client(read_env = True)
Cube.setclient(cli)
Workflow.setclient(cli)
```

```python
hosts = "1"
cores = "1"
threads = "8"
partition = "partition"
display = True
inputfile = "https://www.unidata.ucar.edu/software/netcdf/examples/tos_O1_2001-2002.nc"
inputmeasure = "tos"
```

Startup the computing components on the server-side

```python
Cube.cluster(action = "deploy", host_partition = partition, nhost = hosts, exec_mode = "async")
```

Load CWL-based Ophidia workflow from *example.cwl*

```python
exp = Experiment.load_cwl("example.cwl", "--inputfile " + inputfile + " --inputmeasure " + inputmeasure + " --partition " + partition + " --nthreads " + threads)
```

Before running the workflow, check the definition structure

```python
exp.check(display = display)
```

Link the definition to a workflow object and start its execution on the server side

```python
wf = Workflow(exp)
wf.submit()
```

While the workflow is running monitor its execution

```python
%load_ext autoreload
%autoreload 2
wf.monitor(frequency = 1, iterative = True, display = display)
```

Shut down the computing components on the server-side

```python
Cube.cluster(action = "undeploy", host_partition = partition)
```
